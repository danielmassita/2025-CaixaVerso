```python
import json
```

## Definindo as funções

```python
def criar_banco_vazio(path):
    contas_dict = {}
    salvar_dict_de_contas(path, contas_dict)
    return contas_dict
```

```python
def carregar_contas_como_dict(path):
    try:
        with open(path, 'r') as file:
            conteudo = file.read()
            conteudo_dict = json.loads(conteudo)
        return conteudo_dict
    except FileNotFoundError:
        print(f"Arquivo {path} não encontrado.")
        return {}
    except json.JSONDecodeError:
        print(f"Erro ao decodificar o JSON no arquivo {path}.")
        return {}
```

```python
def salvar_dict_de_contas(path, contas_dict):
    try:
        with open(path, 'w') as file:
            with open(path, 'w') as file:
                conteudo_json = json.dumps(contas_dict)
                file.write(conteudo_json)
    except IOError as e:
        print(f"Erro ao salvar o arquivo {path}: {e}")
```

```python
def cadastrar_novo_cliente(path, id, nome, sobrenome, cpf, saldo = 0):
    contas_dict = carregar_contas_como_dict(path)
    if id in contas_dict:
        print(f"ID {id} já existe. Por favor, escolha outro ID.")
        return False
    contas_dict[id] = {
        'nome': nome,
        'sobrenome': sobrenome,
        'cpf': cpf,
        'saldo': saldo
    }
    salvar_dict_de_contas(path, contas_dict)
    return True
```

```python
def transferir_saldo(path, conta_origem, conta_destino, valor):
    contas_dict = carregar_contas_como_dict(path)

    if conta_origem not in contas_dict or conta_destino not in contas_dict:
        print("Uma ou ambas as contas não existem.")
        return False

    if contas_dict[conta_origem]['saldo'] < valor:
        print("Saldo insuficiente na conta de origem.")
        return False

    contas_dict[conta_origem]['saldo'] -= valor
    contas_dict[conta_destino]['saldo'] += valor

    salvar_dict_de_contas(path, contas_dict)

    return True
```
